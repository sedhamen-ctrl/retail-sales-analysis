```python
import pandas as pd

df = pd.read_csv("/content/dataset/retail_store_sales.csv")

print("Jumlah baris dan kolom:")
print(df.shape)

print("\nNama kolom:")
print(df.columns.tolist())

print("\n5 data pertama:")
display(df.head())
```

```python
# 1. Informasi tipe data dan jumlah data
df.info()

# 2. Jumlah data kosong
print("\n=== DATA KOSONG ===")
print(df.isnull().sum())

# 3. Jumlah data duplikat
print("\n=== DUPLIKAT ===")
print("Jumlah baris duplikat:", df.duplicated().sum())
print("Transaction ID duplikat:", df["Transaction ID"].duplicated().sum())

# 4. Jumlah nilai unik
print("\n=== NILAI UNIK ===")
print(df.nunique())

# 5. Statistik angka
print("\n=== STATISTIK ===")
display(df.describe())
```

```python
# Melihat nilai yang ada pada kolom-kolom kategorikal

for col in ['Category', 'Payment Method', 'Location', 'Discount Applied']:
    print(f"\n=== {col} ===")
    print(df[col].value_counts(dropna=False))
```

```python
# Cek apakah Total Spent = Price Per Unit × Quantity

df['Calculated Total'] = df['Price Per Unit'] * df['Quantity']

print("Jumlah data yang bisa dibandingkan:",
      df[['Total Spent', 'Calculated Total']].dropna().shape[0])

print("\nContoh perbandingan:")
display(
    df[['Price Per Unit', 'Quantity', 'Total Spent', 'Calculated Total']]
    .dropna()
    .head(10)
)
```

```python
pd.crosstab(
    df['Discount Applied'].fillna('Missing'),
    df['Payment Method']
)
```

```python
pd.crosstab(
    df['Discount Applied'].fillna('Missing'),
    df['Location']
)
```

```python
df.groupby('Discount Applied', dropna=False)[
    ['Price Per Unit', 'Quantity', 'Total Spent']
].mean()
```

```python
print("Missing berdasarkan kolom:")
display(df.isnull().sum())

print("\nPola missing:")
display(
    df.isnull()
      .groupby(['Item', 'Price Per Unit', 'Quantity', 'Total Spent'])
      .size()
      .sort_values(ascending=False)
      .head(15)
)
```

```python
# Cek apakah Category + Price Per Unit bisa menentukan Item
item_mapping = (
    df.dropna(subset=['Item', 'Price Per Unit'])
      .groupby(['Category', 'Price Per Unit'])['Item']
      .nunique()
      .reset_index(name='Jumlah Item')
)

print("Kombinasi Category + Price yang hanya punya 1 Item:")
display(item_mapping[item_mapping['Jumlah Item'] == 1].head(20))

print("\nKombinasi Category + Price yang punya lebih dari 1 Item:")
display(item_mapping[item_mapping['Jumlah Item'] > 1].head(20))
```

```python
print("Missing Item berdasarkan Category:")

missing_item = (
    df[df['Item'].isna()]
    .groupby('Category')
    .size()
    .sort_values(ascending=False)
)

display(missing_item)
```

```python
# Membuat mapping Category + Price -> Item
item_mapping = (
    df.dropna(subset=['Item', 'Price Per Unit'])
      .groupby(['Category', 'Price Per Unit'])['Item']
      .first()
)

# Cari baris yang Item-nya kosong tetapi Category dan Price tersedia
missing_item = df[
    df['Item'].isna() &
    df['Category'].notna() &
    df['Price Per Unit'].notna()
].copy()

# Coba cari Item berdasarkan Category + Price
missing_item['Predicted Item'] = missing_item.apply(
    lambda row: item_mapping.get(
        (row['Category'], row['Price Per Unit'])
    ),
    axis=1
)

print("Total Item missing:", df['Item'].isna().sum())
print("Bisa direcover:", missing_item['Predicted Item'].notna().sum())
print("Tidak bisa direcover:", missing_item['Predicted Item'].isna().sum())

display(missing_item.head(10))
```

```python
# Validasi mapping Item berdasarkan Category + Price

validation = df.dropna(subset=['Item', 'Price Per Unit']).copy()

# Harga yang seharusnya dimiliki setiap Item
item_price_check = (
    validation
    .groupby(['Category', 'Item'])['Price Per Unit']
    .nunique()
    .reset_index(name='Jumlah Harga')
)

print("Item yang memiliki lebih dari 1 harga:")
display(item_price_check[item_price_check['Jumlah Harga'] > 1])
```

```python
# Isi Item yang kosong menggunakan hasil prediksi
df.loc[df['Item'].isna(), 'Item'] = missing_item['Predicted Item']

print("Sisa Item missing:", df['Item'].isna().sum())
```

```python
# Cek apakah setiap Item memiliki satu harga yang konsisten

price_check = (
    df.dropna(subset=['Item', 'Price Per Unit'])
      .groupby('Item')['Price Per Unit']
      .nunique()
)

print("Item dengan lebih dari 1 harga:")
display(price_check[price_check > 1])

print("\nJumlah Item yang memiliki harga konsisten:",
      (price_check == 1).sum())
```

```python
# Buat mapping Item -> Price Per Unit
price_mapping = (
    df.dropna(subset=['Item', 'Price Per Unit'])
      .groupby('Item')['Price Per Unit']
      .first()
)

# Isi Price Per Unit yang kosong berdasarkan Item
df['Price Per Unit'] = df.apply(
    lambda row: price_mapping.get(row['Item'])
    if pd.isna(row['Price Per Unit']) else row['Price Per Unit'],
    axis=1
)

print("Sisa Price Per Unit yang missing:",
      df['Price Per Unit'].isna().sum())
```

```python
# Cek pola missing Quantity dan Total Spent
quantity_total_check = df[
    df['Quantity'].isna() | df['Total Spent'].isna()
][['Item', 'Price Per Unit', 'Quantity', 'Total Spent']]

display(quantity_total_check.head(20))

print("Jumlah baris yang Quantity missing:",
      df['Quantity'].isna().sum())

print("Jumlah baris yang Total Spent missing:",
      df['Total Spent'].isna().sum())

print("\nPola missing:")
display(
    df[['Price Per Unit', 'Quantity', 'Total Spent']]
    .isna()
    .value_counts()
    .reset_index(name='Jumlah')
)
```

```python
# Distribusi Quantity berdasarkan Item

quantity_pattern = (
    df.dropna(subset=['Item', 'Quantity'])
      .groupby('Item')['Quantity']
      .agg(['count', 'mean', 'median', 'min', 'max'])
      .sort_values('count', ascending=False)
)

display(quantity_pattern.head(20))
```

```python
quantity_variation = (
    df.dropna(subset=['Item', 'Quantity'])
      .groupby('Item')['Quantity']
      .nunique()
)

print("Jumlah Item:", len(quantity_variation))
print("\nDistribusi jumlah variasi Quantity:")
display(quantity_variation.value_counts().sort_index())
```

```python
# Simpan data asli
df_raw = df.copy()

# Dataset untuk analisis
df_clean = df.dropna(
    subset=['Price Per Unit', 'Quantity', 'Total Spent']
).copy()

print("Data awal :", len(df_raw))
print("Data setelah cleaning :", len(df_clean))
print("Data yang dikeluarkan :", len(df_raw) - len(df_clean))

print("\nMissing setelah cleaning:")
display(df_clean.isnull().sum())
```

```python
# Distribusi Discount Applied
print("Distribusi Discount Applied:")

display(
    df_clean['Discount Applied']
    .value_counts(dropna=False)
    .rename_axis('Discount Applied')
    .reset_index(name='Jumlah')
)
```

```python
# Bandingkan rata-rata transaksi berdasarkan status discount
display(
    df_clean.groupby(
        df_clean['Discount Applied'].fillna('Missing')
    )[['Price Per Unit', 'Quantity', 'Total Spent']]
    .mean()
)
```

```python
# Ubah missing Discount Applied menjadi Unknown
df_clean['Discount Applied'] = (
    df_clean['Discount Applied']
    .fillna('Unknown')
    .astype(str)
)

# Cek hasil
print("Distribusi Discount Applied setelah cleaning:")

display(
    df_clean['Discount Applied']
    .value_counts()
    .rename_axis('Discount Applied')
    .reset_index(name='Jumlah')
)
```

```python
print("=== TIPE DATA ===")
print(df_clean.dtypes)
```

```python
df_clean['Transaction Date'] = pd.to_datetime(
    df_clean['Transaction Date'],
    errors='coerce'
)

print(df_clean.dtypes)
```

```python
print("Tanggal invalid:", df_clean['Transaction Date'].isna().sum())
```

```python
print("Tanggal paling awal :", df_clean['Transaction Date'].min())
print("Tanggal paling akhir:", df_clean['Transaction Date'].max())
```

```python
print("=== CEK PRICE ===")
print("Harga <= 0:", (df_clean['Price Per Unit'] <= 0).sum())

print("\n=== CEK QUANTITY ===")
print("Quantity <= 0:", (df_clean['Quantity'] <= 0).sum())
```

```python
df_clean['Calculated Total'] = (
    df_clean['Price Per Unit'] * df_clean['Quantity']
)

df_clean['Total Difference'] = (
    df_clean['Total Spent'] - df_clean['Calculated Total']
)

print(
    "Total transaksi yang tidak sesuai:",
    (df_clean['Total Difference'] != 0).sum()
)
```

```python
df_clean = df_clean.drop(columns=['Calculated Total'])
```

```python
print(df_clean.shape)
display(df_clean.head())
```

```python
df_clean['Year'] = df_clean['Transaction Date'].dt.year

df_clean['Month'] = df_clean['Transaction Date'].dt.month

df_clean['Month Name'] = df_clean['Transaction Date'].dt.month_name()

df_clean['Year-Month'] = (
    df_clean['Transaction Date']
    .dt.to_period('M')
    .astype(str)
)

df_clean['Day of Week'] = (
    df_clean['Transaction Date']
    .dt.day_name()
)
```

```python
df_clean['Day Type'] = df_clean['Transaction Date'].dt.dayofweek.apply(
    lambda x: 'Weekend' if x >= 5 else 'Weekday'
)
```

```python
print(df_clean.columns.tolist())
print(df_clean.shape)
display(df_clean.head())
```

```python
df_clean = df_clean.drop(columns=['Total Difference'])
```

```python
print("=== BUSINESS OVERVIEW ===")

total_revenue = df_clean['Total Spent'].sum()
total_transactions = df_clean['Transaction ID'].nunique()
total_customers = df_clean['Customer ID'].nunique()
total_quantity = df_clean['Quantity'].sum()
average_transaction = df_clean['Total Spent'].mean()

print(f"Total Revenue       : ${total_revenue:,.2f}")
print(f"Total Transactions  : {total_transactions:,}")
print(f"Total Customers     : {total_customers:,}")
print(f"Total Quantity Sold : {total_quantity:,.0f}")
print(f"Average Transaction : ${average_transaction:,.2f}")
```

```python
# ==========================================
# EDA 2 - CATEGORY & PRODUCT ANALYSIS
# ==========================================

# 1. Analisis berdasarkan kategori
category_analysis = (
    df_clean.groupby('Category')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .sort_values('Revenue', ascending=False)
    .reset_index()
)

print("=== ANALISIS KATEGORI ===")
display(category_analysis)


# 2. Top 10 produk berdasarkan quantity
top_products_quantity = (
    df_clean.groupby('Item')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .sort_values('Quantity_Sold', ascending=False)
    .head(10)
    .reset_index()
)

print("=== TOP 10 PRODUK BERDASARKAN QUANTITY ===")
display(top_products_quantity)


# 3. Top 10 produk berdasarkan revenue
top_products_revenue = (
    df_clean.groupby('Item')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .sort_values('Revenue', ascending=False)
    .head(10)
    .reset_index()
)

print("=== TOP 10 PRODUK BERDASARKAN REVENUE ===")
display(top_products_revenue)
```

```python
# ==========================================
# EDA 3 - TIME ANALYSIS
# ==========================================

monthly_analysis = (
    df_clean.groupby('Year-Month')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .reset_index()
    .sort_values('Year-Month')
)

print("=== MONTHLY SALES ANALYSIS ===")
display(monthly_analysis)

print("\n=== BULAN DENGAN REVENUE TERTINGGI ===")
display(
    monthly_analysis.loc[
        monthly_analysis['Revenue'].idxmax()
    ]
)

print("\n=== BULAN DENGAN REVENUE TERENDAH ===")
display(
    monthly_analysis.loc[
        monthly_analysis['Revenue'].idxmin()
    ]
)
```

```python
# ==========================================
# EDA 4 - LOCATION, PAYMENT & DISCOUNT
# ==========================================

# 1. ANALISIS LOCATION
location_analysis = (
    df_clean.groupby('Location')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .reset_index()
    .sort_values('Revenue', ascending=False)
)

print("=== LOCATION ANALYSIS ===")
display(location_analysis)


# 2. ANALISIS PAYMENT METHOD
payment_analysis = (
    df_clean.groupby('Payment Method')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .reset_index()
    .sort_values('Revenue', ascending=False)
)

print("\n=== PAYMENT METHOD ANALYSIS ===")
display(payment_analysis)


# 3. ANALISIS DISCOUNT
discount_analysis = (
    df_clean.groupby('Discount Applied')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum'),
        Avg_Transaction=('Total Spent', 'mean')
    )
    .reset_index()
    .sort_values('Revenue', ascending=False)
)

print("\n=== DISCOUNT ANALYSIS ===")
display(discount_analysis)
```

```python
# ==========================================
# EDA 5 - CROSS ANALYSIS
# ==========================================

# 1. Revenue berdasarkan Category + Location
category_location = (
    df_clean.groupby(['Category', 'Location'])
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .reset_index()
    .sort_values('Revenue', ascending=False)
)

print("=== CATEGORY × LOCATION ===")
display(category_location)


# 2. Revenue berdasarkan Category + Discount
category_discount = (
    df_clean.groupby(['Category', 'Discount Applied'])
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Revenue=('Total Spent', 'sum'),
        Avg_Transaction=('Total Spent', 'mean')
    )
    .reset_index()
    .sort_values('Revenue', ascending=False)
)

print("\n=== CATEGORY × DISCOUNT ===")
display(category_discount)


# 3. Top product berdasarkan revenue dan quantity
product_performance = (
    df_clean.groupby(['Item', 'Category'])
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity_Sold=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .reset_index()
)

# Tambahkan ranking
product_performance['Revenue_Rank'] = (
    product_performance['Revenue']
    .rank(method='dense', ascending=False)
)

product_performance['Quantity_Rank'] = (
    product_performance['Quantity_Sold']
    .rank(method='dense', ascending=False)
)

print("\n=== PRODUCT PERFORMANCE ===")
display(
    product_performance
    .sort_values('Revenue', ascending=False)
    .head(15)
)
```

```python
# ============================================================
# RETAIL SALES ANALYSIS - PROFESSIONAL EDA VISUALIZATION
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# ------------------------------------------------------------
# FORMATTER
# ------------------------------------------------------------

def rupiah(x, pos):
    return f'${x/1000:.0f}K'

currency_formatter = FuncFormatter(rupiah)


# ============================================================
# 1. MONTHLY REVENUE TREND
# ============================================================

monthly = (
    df_clean.groupby('Year-Month')
    .agg(
        Revenue=('Total Spent', 'sum'),
        Transactions=('Transaction ID', 'nunique')
    )
    .reset_index()
)

# Cari bulan tertinggi & terendah
max_month = monthly.loc[monthly['Revenue'].idxmax()]
min_month = monthly.loc[monthly['Revenue'].idxmin()]

plt.figure(figsize=(15, 6))

plt.plot(
    monthly['Year-Month'],
    monthly['Revenue'],
    marker='o',
    linewidth=2
)

# Highlight highest
plt.scatter(
    max_month['Year-Month'],
    max_month['Revenue'],
    s=100,
    zorder=5
)

plt.annotate(
    f"Highest\n{max_month['Year-Month']}\n${max_month['Revenue']:,.0f}",
    xy=(max_month['Year-Month'], max_month['Revenue']),
    xytext=(0, 25),
    textcoords='offset points',
    ha='center',
    fontsize=10
)

plt.title(
    'Monthly Revenue Trend',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Month')
plt.ylabel('Revenue')

plt.gca().yaxis.set_major_formatter(currency_formatter)

plt.xticks(
    range(0, len(monthly), 3),
    monthly['Year-Month'].iloc[::3],
    rotation=45
)

plt.grid(axis='y', alpha=0.25)

plt.tight_layout()
plt.show()


# ============================================================
# 2. REVENUE BY CATEGORY
# ============================================================

category = (
    df_clean.groupby('Category')
    .agg(
        Revenue=('Total Spent', 'sum'),
        Quantity=('Quantity', 'sum'),
        Transactions=('Transaction ID', 'nunique')
    )
    .sort_values('Revenue', ascending=True)
)

plt.figure(figsize=(11, 7))

bars = plt.barh(
    category.index,
    category['Revenue']
)

plt.title(
    'Revenue by Product Category',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Revenue')
plt.ylabel('Category')

plt.gca().xaxis.set_major_formatter(currency_formatter)

# Label nilai
for bar in bars:
    width = bar.get_width()

    plt.text(
        width + 2000,
        bar.get_y() + bar.get_height()/2,
        f'${width:,.0f}',
        va='center',
        fontsize=9
    )

plt.grid(axis='x', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 3. TOP 10 PRODUCTS BY REVENUE
# ============================================================

top_products = (
    df_clean.groupby('Item')
    .agg(
        Revenue=('Total Spent', 'sum'),
        Quantity=('Quantity', 'sum')
    )
    .sort_values('Revenue', ascending=False)
    .head(10)
    .sort_values('Revenue')
)

plt.figure(figsize=(11, 7))

bars = plt.barh(
    top_products.index,
    top_products['Revenue']
)

plt.title(
    'Top 10 Products by Revenue',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Revenue')
plt.ylabel('Product')

plt.gca().xaxis.set_major_formatter(currency_formatter)

for bar in bars:
    width = bar.get_width()

    plt.text(
        width + 300,
        bar.get_y() + bar.get_height()/2,
        f'${width:,.0f}',
        va='center',
        fontsize=9
    )

plt.grid(axis='x', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 4. TOP 10 PRODUCTS BY QUANTITY
# ============================================================

top_quantity = (
    df_clean.groupby('Item')
    .agg(
        Quantity=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .sort_values('Quantity', ascending=False)
    .head(10)
    .sort_values('Quantity')
)

plt.figure(figsize=(11, 7))

bars = plt.barh(
    top_quantity.index,
    top_quantity['Quantity']
)

plt.title(
    'Top 10 Products by Quantity Sold',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Units Sold')
plt.ylabel('Product')

for bar in bars:
    width = bar.get_width()

    plt.text(
        width + 10,
        bar.get_y() + bar.get_height()/2,
        f'{width:,.0f}',
        va='center',
        fontsize=9
    )

plt.grid(axis='x', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 5. ONLINE VS IN-STORE
# ============================================================

location = (
    df_clean.groupby('Location')
    .agg(
        Revenue=('Total Spent', 'sum'),
        Transactions=('Transaction ID', 'nunique'),
        Quantity=('Quantity', 'sum')
    )
    .reset_index()
)

plt.figure(figsize=(8, 6))

bars = plt.bar(
    location['Location'],
    location['Revenue']
)

plt.title(
    'Revenue by Sales Channel',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Sales Channel')
plt.ylabel('Revenue')

plt.gca().yaxis.set_major_formatter(currency_formatter)

for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 10000,
        f'${height:,.0f}',
        ha='center',
        fontweight='bold'
    )

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 6. PAYMENT METHOD
# ============================================================

payment = (
    df_clean.groupby('Payment Method')['Total Spent']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 6))

bars = plt.bar(
    payment.index,
    payment.values
)

plt.title(
    'Revenue by Payment Method',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Payment Method')
plt.ylabel('Revenue')

plt.gca().yaxis.set_major_formatter(currency_formatter)

for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 5000,
        f'${height:,.0f}',
        ha='center',
        fontweight='bold'
    )

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 7. CATEGORY: REVENUE VS QUANTITY
# ============================================================

category_scatter = (
    df_clean.groupby('Category')
    .agg(
        Revenue=('Total Spent', 'sum'),
        Quantity=('Quantity', 'sum')
    )
    .reset_index()
)

plt.figure(figsize=(10, 7))

plt.scatter(
    category_scatter['Quantity'],
    category_scatter['Revenue'],
    s=120
)

for _, row in category_scatter.iterrows():

    plt.annotate(
        row['Category'],
        (
            row['Quantity'],
            row['Revenue']
        ),
        xytext=(5, 5),
        textcoords='offset points',
        fontsize=9
    )

plt.title(
    'Category Performance: Quantity vs Revenue',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Quantity Sold')
plt.ylabel('Revenue')

plt.gca().yaxis.set_major_formatter(currency_formatter)

plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 8. DISCOUNT ANALYSIS
# ============================================================

discount = (
    df_clean.groupby('Discount Applied')
    .agg(
        Revenue=('Total Spent', 'sum'),
        Transactions=('Transaction ID', 'nunique'),
        Avg_Transaction=('Total Spent', 'mean')
    )
    .reset_index()
)

plt.figure(figsize=(8, 6))

bars = plt.bar(
    discount['Discount Applied'].astype(str),
    discount['Revenue']
)

plt.title(
    'Revenue by Discount Status',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Discount Status')
plt.ylabel('Revenue')

plt.gca().yaxis.set_major_formatter(currency_formatter)

for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 5000,
        f'${height:,.0f}',
        ha='center',
        fontweight='bold'
    )

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# EDA LANJUTAN - RETAIL SALES ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. DISTRIBUSI TOTAL SPENT
# ============================================================

print("=== DISTRIBUSI TOTAL SPENT ===")

print(df_clean['Total Spent'].describe())

plt.figure(figsize=(10, 6))

plt.hist(
    df_clean['Total Spent'],
    bins=30,
    edgecolor='black'
)

plt.title(
    'Distribution of Transaction Value',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Total Spent')
plt.ylabel('Number of Transactions')

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 2. DISTRIBUSI QUANTITY
# ============================================================

print("=== DISTRIBUSI QUANTITY ===")

print(df_clean['Quantity'].describe())

plt.figure(figsize=(9, 5))

quantity_counts = df_clean['Quantity'].value_counts().sort_index()

plt.bar(
    quantity_counts.index.astype(str),
    quantity_counts.values
)

plt.title(
    'Distribution of Quantity per Transaction',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Quantity')
plt.ylabel('Number of Transactions')

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 3. OUTLIER TOTAL SPENT - IQR METHOD
# ============================================================

print("=== OUTLIER TOTAL SPENT ===")

Q1 = df_clean['Total Spent'].quantile(0.25)
Q3 = df_clean['Total Spent'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean['Total Spent'] < lower_bound) |
    (df_clean['Total Spent'] > upper_bound)
]

print(f"Q1              : {Q1:.2f}")
print(f"Q3              : {Q3:.2f}")
print(f"IQR             : {IQR:.2f}")
print(f"Lower Bound     : {lower_bound:.2f}")
print(f"Upper Bound     : {upper_bound:.2f}")
print(f"Jumlah Outlier  : {len(outliers)}")

display(
    outliers[
        [
            'Transaction ID',
            'Customer ID',
            'Category',
            'Item',
            'Price Per Unit',
            'Quantity',
            'Total Spent'
        ]
    ]
    .sort_values('Total Spent', ascending=False)
    .head(20)
)


# ============================================================
# 4. BOXPLOT TOTAL SPENT
# ============================================================

plt.figure(figsize=(10, 4))

plt.boxplot(
    df_clean['Total Spent'],
    vert=False
)

plt.title(
    'Boxplot of Transaction Value',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Total Spent')

plt.grid(axis='x', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 5. CUSTOMER PERFORMANCE
# ============================================================

customer_analysis = (
    df_clean.groupby('Customer ID')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Total_Spent=('Total Spent', 'sum'),
        Quantity=('Quantity', 'sum'),
        Avg_Transaction=('Total Spent', 'mean')
    )
    .sort_values('Total_Spent', ascending=False)
    .reset_index()
)

print("=== TOP 10 CUSTOMER BERDASARKAN TOTAL SPENDING ===")

display(
    customer_analysis.head(10)
)


# ============================================================
# 6. CUSTOMER PALING SERING TRANSAKSI
# ============================================================

top_customers_transaction = (
    customer_analysis
    .sort_values('Transactions', ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_customers_transaction['Customer ID'],
    top_customers_transaction['Transactions']
)

plt.gca().invert_yaxis()

plt.title(
    'Top 10 Customers by Number of Transactions',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Number of Transactions')
plt.ylabel('Customer')

plt.grid(axis='x', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 7. CUSTOMER TOTAL SPENDING
# ============================================================

top_customers_spending = (
    customer_analysis
    .sort_values('Total_Spent', ascending=False)
    .head(10)
    .sort_values('Total_Spent')
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_customers_spending['Customer ID'],
    top_customers_spending['Total_Spent']
)

plt.title(
    'Top 10 Customers by Total Spending',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Total Spending')
plt.ylabel('Customer')

plt.grid(axis='x', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 8. REPEAT CUSTOMER ANALYSIS
# ============================================================

customer_analysis['Customer Type'] = np.where(
    customer_analysis['Transactions'] > 1,
    'Repeat Customer',
    'One-time Customer'
)

customer_type = (
    customer_analysis['Customer Type']
    .value_counts()
)

print("=== CUSTOMER TYPE ===")
print(customer_type)

plt.figure(figsize=(7, 5))

plt.bar(
    customer_type.index,
    customer_type.values
)

plt.title(
    'Customer Type Distribution',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Customer Type')
plt.ylabel('Number of Customers')

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 9. WEEKDAY VS WEEKEND
# ============================================================

day_analysis = (
    df_clean.groupby('Day Type')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum'),
        Avg_Transaction=('Total Spent', 'mean')
    )
    .reset_index()
)

print("=== WEEKDAY VS WEEKEND ===")

display(day_analysis)


# Revenue
plt.figure(figsize=(7, 5))

plt.bar(
    day_analysis['Day Type'],
    day_analysis['Revenue']
)

plt.title(
    'Revenue: Weekday vs Weekend',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Day Type')
plt.ylabel('Revenue')

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 10. REVENUE PER DAY OF WEEK
# ============================================================

weekday_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

weekday_analysis = (
    df_clean.groupby('Day of Week')['Total Spent']
    .sum()
    .reindex(weekday_order)
)

plt.figure(figsize=(10, 5))

plt.plot(
    weekday_analysis.index,
    weekday_analysis.values,
    marker='o',
    linewidth=2
)

plt.title(
    'Revenue by Day of Week',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Day')
plt.ylabel('Revenue')

plt.xticks(rotation=30)

plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 11. PRICE VS QUANTITY
# ============================================================

price_quantity = (
    df_clean.groupby('Price Per Unit')
    .agg(
        Avg_Quantity=('Quantity', 'mean'),
        Total_Quantity=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum')
    )
    .reset_index()
)

plt.figure(figsize=(10, 6))

plt.scatter(
    price_quantity['Price Per Unit'],
    price_quantity['Avg_Quantity'],
    s=80
)

plt.title(
    'Price vs Average Quantity per Transaction',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Price Per Unit')
plt.ylabel('Average Quantity')

plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 12. QUANTITY VS TOTAL SPENT
# ============================================================

plt.figure(figsize=(10, 6))

plt.scatter(
    df_clean['Quantity'],
    df_clean['Total Spent'],
    alpha=0.25
)

plt.title(
    'Quantity vs Transaction Value',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Quantity')
plt.ylabel('Total Spent')

plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 13. KORELASI NUMERIK
# ============================================================

numeric_cols = [
    'Price Per Unit',
    'Quantity',
    'Total Spent'
]

correlation = df_clean[numeric_cols].corr()

print("=== CORRELATION MATRIX ===")

display(correlation)


# Visualisasi correlation sederhana

plt.figure(figsize=(7, 5))

plt.imshow(
    correlation,
    interpolation='nearest'
)

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=30
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title(
    'Correlation Matrix',
    fontsize=15,
    fontweight='bold'
)

# Tambahkan nilai korelasi
for i in range(len(correlation)):
    for j in range(len(correlation.columns)):
        plt.text(
            j,
            i,
            f"{correlation.iloc[i, j]:.2f}",
            ha='center',
            va='center'
        )

plt.tight_layout()
plt.show()


# ============================================================
# 14. REVENUE PER TAHUN
# ============================================================

year_analysis = (
    df_clean.groupby('Year')
    .agg(
        Transactions=('Transaction ID', 'nunique'),
        Quantity=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum'),
        Avg_Transaction=('Total Spent', 'mean')
    )
    .reset_index()
)

print("=== YEARLY PERFORMANCE ===")

display(year_analysis)


# ============================================================
# 15. PERTUMBUHAN REVENUE TAHUNAN
# ============================================================

year_analysis['Revenue Growth (%)'] = (
    year_analysis['Revenue']
    .pct_change()
    * 100
)

print("=== REVENUE GROWTH ===")

display(
    year_analysis[
        [
            'Year',
            'Revenue',
            'Revenue Growth (%)'
        ]
    ]
)


# ============================================================
# 16. REVENUE TAHUNAN
# ============================================================

plt.figure(figsize=(9, 5))

plt.bar(
    year_analysis['Year'].astype(str),
    year_analysis['Revenue']
)

plt.title(
    'Annual Revenue',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Year')
plt.ylabel('Revenue')

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 17. AVERAGE TRANSACTION PER CATEGORY
# ============================================================

category_aov = (
    df_clean.groupby('Category')['Total Spent']
    .mean()
    .sort_values(ascending=True)
)

plt.figure(figsize=(10, 6))

plt.barh(
    category_aov.index,
    category_aov.values
)

plt.title(
    'Average Transaction Value by Category',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Average Transaction Value')
plt.ylabel('Category')

plt.grid(axis='x', alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 18. PRODUK: REVENUE VS QUANTITY
# ============================================================

product_analysis = (
    df_clean.groupby(['Item', 'Category'])
    .agg(
        Quantity=('Quantity', 'sum'),
        Revenue=('Total Spent', 'sum'),
        Transactions=('Transaction ID', 'nunique'),
        Avg_Price=('Price Per Unit', 'mean')
    )
    .reset_index()
)

plt.figure(figsize=(11, 7))

plt.scatter(
    product_analysis['Quantity'],
    product_analysis['Revenue'],
    s=60,
    alpha=0.6
)

plt.title(
    'Product Performance: Quantity vs Revenue',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Total Quantity Sold')
plt.ylabel('Total Revenue')

plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()


# ============================================================
# 19. PRODUK DENGAN REVENUE TINGGI TAPI QUANTITY RENDAH
# ============================================================

product_analysis['Revenue_per_Unit'] = (
    product_analysis['Revenue']
    / product_analysis['Quantity']
)

high_value_products = (
    product_analysis
    .sort_values('Revenue_per_Unit', ascending=False)
    .head(10)
)

print("=== TOP 10 PRODUCT BERDASARKAN REVENUE PER UNIT ===")

display(
    high_value_products[
        [
            'Item',
            'Category',
            'Quantity',
            'Revenue',
            'Revenue_per_Unit'
        ]
    ]
)


# ============================================================
# 20. SUMMARY EDA
# ============================================================

print("=" * 60)
print("SUMMARY EDA")
print("=" * 60)

print(f"Total Revenue          : ${df_clean['Total Spent'].sum():,.2f}")
print(f"Total Transactions     : {df_clean['Transaction ID'].nunique():,}")
print(f"Total Customers        : {df_clean['Customer ID'].nunique():,}")
print(f"Total Quantity Sold    : {df_clean['Quantity'].sum():,}")
print(f"Average Transaction    : ${df_clean['Total Spent'].mean():,.2f}")

print("\nKategori Revenue Tertinggi:")
print(
    df_clean.groupby('Category')['Total Spent']
    .sum()
    .idxmax()
)

print("\nProduk Revenue Tertinggi:")
print(
    df_clean.groupby('Item')['Total Spent']
    .sum()
    .idxmax()
)

print("\nProduk Quantity Tertinggi:")
print(
    df_clean.groupby('Item')['Quantity']
    .sum()
    .idxmax()
)

print("\nChannel Revenue Tertinggi:")
print(
    df_clean.groupby('Location')['Total Spent']
    .sum()
    .idxmax()
)
```
